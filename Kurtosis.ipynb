```python
import numpy as np
import pandas as pd
from scipy.stats import kurtosis as k
data=pd.read_csv("Store.csv",encoding="latin-1")
print(data)
```

```python
from scipy.stats import kurtosis as k

r=data["Profit"].astype(int)

d=k(r, fisher=True)
print("kurtosis value is ",d)

if d < 0:
    print("Platykurtic → Light tail, low variation")
else:
    print("Not Platykurtic")

```

```python
plt.plot(d, color="pink")
plt.title("Profit or Loss Distribution")
plt.xlabel("Profit Values")
plt.ylabel("Frequency")
plt.axhline(d, linestyle="--", linewidth=3)
plt.show()
```
