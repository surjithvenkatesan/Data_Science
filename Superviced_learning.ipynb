```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import seaborn as sns
import matplotlib.pyplot as plt


#LinearRegression model practice  with data of Land(sqft) and price prediction

data=pd.read_csv("data.csv")
x=np.array(data['sqft_living']).reshape(-1,1)
y=data['price']

fa= LinearRegression()
fa.fit(x,y)

new_pre=fa.predict(x)

prediction = fa.predict([[1710]])
print("Predicted price for :", prediction[0])

```
