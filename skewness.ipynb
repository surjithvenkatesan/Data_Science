```python
import numpy as np
import pandas as pd
file=pd.read_csv(r"/content/Car Sales.xlsx - car_data.csv")
c=(file["Annual Income"])
print(c.isnull().sum())
```

```python
print(c)
```

```python
from scipy.stats import skew
sk=skew(c)
print(sk)
if sk>0:
    print("positive skew")
elif sk<0:
    print("negative skew")
else:
    print("normal skew")
```

```python
import matplotlib.pyplot as plt
plt.hist(sk, color="blue", bins=30)
plt.title("profit distribution")
plt.xlabel("Profit")
plt.show()
```
